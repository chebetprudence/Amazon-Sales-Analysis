# Amazon Sales Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('amazon.csv')
df.head()
```

```python
# get a full picture of the data 

print("Shape of the dataset:", df.shape)
print("\n ---- Data Types ----")
print(df.dtypes)
print("\n ---- Missing Values ----")
print(df.isnull().sum())
print("\n ---- Duplicates ----")
print("Full duplicate rows: ", df.duplicated().sum())
print("Duplicate product_ids: ", df['product_id'].duplicated().sum())
```

```python
# Inspecting missing values in the rating_count column

print(df[df['rating_count'].isnull()][['product_id', 'product_name', 'rating', 'rating_count']])
```

```python
# fill the missing rating_count values with 0, indicates it has no ratings yet. dropping it we'd  be loosing a valid product that already has reviews.

df['rating_count'] = df['rating_count'].fillna(0)

print("missing values in rating_count:", df['rating_count'].isnull().sum())

# confirming if the two rows got filled with 0
df.loc[[282, 324], ['product_name', 'rating', 'rating_count']]
```

```python
df[['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']].head()
```

```python
# clean and convert numeric columns

df['discounted_price'] = df['discounted_price'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '', regex=False).astype(float)

df['rating_count'] = df['rating_count'].str.replace(',', '', regex=False).astype(float)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce') # errors='coerce' turns any non-numeric values into NaN, which we can handle later

print("\n ---- Data Types After Cleaning ----")
print(df[['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']].dtypes)

print("---- sample of cleaned data ----")
print(df[['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']].head())

print("\n ---- Missing Values After Cleaning ----")
print(df['rating'].isnull().sum())
```

```python
# find the bad rating row
df[df['rating'].isnull()][['product_id', 'product_name', 'rating', 'rating_count']]
```

```python
# fill the missing rating values with 0
df['rating'] =df['rating'].fillna(0)

# Verify
print("NaNs in rating:", df['rating'].isnull().sum())
print(df.loc[1279, ['product_name', 'rating', 'rating_count']])
```

```python
# Fix: re-fill the 2 missing rating_count values
df['rating_count'] = df['rating_count'].fillna(0)

# Verify
print("Missing in rating_count:", df['rating_count'].isnull().sum())
print("Missing in sub_sub_category:", df['sub_sub_category'].isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Create products dataframe, one row per unique product
products_df = df.drop_duplicates(subset='product_id')[[ 
    'product_id', 'product_name', 'main_category', 'sub_category', 
    'sub_sub_category', 'discounted_price', 'actual_price', 
    'discount_percentage', 'discount_amount', 'rating', 
    'rating_count', 'about_product', 'img_link', 'product_link'
]]

# Create reviews dataframe
reviews_df = df[['review_id', 'product_id', 'user_id', 'user_name', 
                 'review_title', 'review_content']].copy()

print("Products table shape:", products_df.shape)
print("Reviews table shape:", reviews_df.shape)
```

```python
# Check mismatched rows
reviews_df['uid_count'] = reviews_df['user_id'].str.split(',').str.len()
reviews_df['uname_count'] = reviews_df['user_name'].str.split(',').str.len()

mismatched = reviews_df[reviews_df['uid_count'] != reviews_df['uname_count']]
print("Mismatched rows:", len(mismatched))
print(mismatched[['user_id', 'user_name', 'uid_count', 'uname_count']].head())
```

```python
# Drop the helper columns first
reviews_df = reviews_df.drop(columns=['uid_count', 'uname_count'])

# Keep only matched rows
matched_reviews = reviews_df[
    reviews_df['user_id'].str.split(',').str.len() == 
    reviews_df['user_name'].str.split(',').str.len()
].copy()

print("Rows kept:", len(matched_reviews))
print("Rows dropped:", len(reviews_df) - len(matched_reviews))

# Now explode safely
reviews_expanded = matched_reviews.copy()
reviews_expanded['user_id'] = matched_reviews['user_id'].str.split(',')
reviews_expanded['user_name'] = matched_reviews['user_name'].str.split(',')
reviews_expanded = reviews_expanded.explode(['user_id', 'user_name'])

# Clean whitespace
reviews_expanded['user_id'] = reviews_expanded['user_id'].str.strip()
reviews_expanded['user_name'] = reviews_expanded['user_name'].str.strip()

# Remove anonymous usernames
fake_users = ['Amazon Customer', 'Placeholder', 'Kindle Customer']
reviews_expanded = reviews_expanded[~reviews_expanded['user_name'].isin(fake_users)]

print("Total unique users:", reviews_expanded['user_id'].nunique())
print(reviews_expanded['user_name'].value_counts().head(10))
```

```python
# Remove placeholder/anonymous usernames
fake_users = ['Amazon Customer', 'Placeholder', 'Kindle Customer']
reviews_cleaned = reviews_expanded[~reviews_expanded['user_name'].isin(fake_users)]

# Recheck
print("Total unique real users:", reviews_cleaned['user_id'].nunique())
print("\nTop 10 most active users:")
print(reviews_cleaned['user_name'].value_counts().head(10))
```

```python
users_df = (reviews_cleaned[['user_id', 'user_name', 'product_id']].drop_duplicates().copy())

user_summary_df = reviews_cleaned.groupby(['user_id', 'user_name']).agg(
    total_reviews=('user_id', 'count'),
).reset_index().sort_values(by='total_reviews', ascending=False)

print(user_summary_df.head(10))
print("Shape: ", user_summary_df.shape)

```

```python
# Simple sentiment analysis using keywords

positive_words = ['good', 'great', 'excellent', 'amazing', 'best', 'love', 
                  'perfect', 'awesome', 'fantastic', 'satisfied', 'happy',
                  'worth', 'recommend', 'nice', 'quality', 'fast']

negative_words = ['bad', 'worst', 'terrible', 'poor', 'waste', 'cheap', 
                  'broken', 'disappointed', 'return', 'refund', 'slow',
                  'damage', 'fake', 'useless', 'horrible', 'not working']

def get_sentiment(text):
    if not isinstance(text, str):
        return 'neutral'
    text = text.lower()
    pos = sum(word in text for word in positive_words)
    neg = sum(word in text for word in negative_words)
    if pos > neg:
        return 'positive'
    elif neg > pos:
        return 'negative'
    else:
        return 'neutral'
reviews_expanded['sentiment'] = reviews_expanded['review_content'].apply(get_sentiment)
print(reviews_expanded['sentiment'].value_counts())   
print(reviews_expanded[['sentiment', 'product_id']].head(5))
```

```python
# Most liked review -- highest positive review count

most_liked = reviews_expanded[reviews_expanded['sentiment'] == 'positive']\
    .groupby('product_id')['sentiment'].count()\
        .reset_index()\
            .rename(columns={'sentiment': 'positive_reviews'})\
                .sort_values(by='positive_reviews', ascending=False)\
                    .head(10)
                    
                    
most_disliked = reviews_expanded[reviews_expanded['sentiment'] == 'negative']\
    .groupby('product_id')['sentiment'].count()\
        .reset_index()\
            .rename(columns={'sentiment': 'negative_reviews'})\
                .sort_values(by='negative_reviews', ascending=False)\
                    .head(10)

# Merge with product names
most_liked = most_liked.merge(products_df[['product_id', 'product_name', 'main_category']], on = 'product_id')
most_disliked = most_disliked.merge(products_df[['product_id', 'product_name', 'main_category']], on = 'product_id')

print("--- TOP 10 MOST LIKED PRODUCTS ---")
print(most_liked[['product_name', 'main_category', 'positive_reviews']].to_string())
print("\n--- TOP 10 MOST DISLIKED PRODUCTS ---")
print(most_disliked[['product_name', 'main_category', 'negative_reviews']].to_string())               
        
```

```python
sentiment_summary = reviews_expanded.groupby(['product_id', 'sentiment']).size()\
    .unstack(fill_value=0).reset_index()

# rename columns
sentiment_summary.columns.name = None
# Add missing columns in case some sentiments don't exist

for col in ['positive', 'negative', 'neutral']:
    if col not in sentiment_summary.columns:
        sentiment_summary[col] = 0

sentiment_summary = sentiment_summary.rename(columns={
    'positive': 'positive_reviews',
    'negative': 'negative_reviews',
    'neutral': 'neutral_reviews'
})
# Add overall sentiment label
sentiment_summary['overall_sentiment'] = sentiment_summary.apply(
    lambda x: 'positive' if x['positive_reviews'] > x['negative_reviews'] 
    else ('negative' if x['negative_reviews'] > x['positive_reviews'] 
          else 'neutral'), axis=1
)

print(sentiment_summary.head(10))
print("Shape:", sentiment_summary.shape)
```

```python
# Calculate the correlation coefficient
correlation_coefficient = df["discounted_price"].corr(df["rating"])
corr_actual = df["actual_price"].corr(df["rating"])
corr_discount_pct = df["discount_percentage"].corr(df["rating"])
corr_discount_amt = df["discount_amount"].corr(df["rating"])

print(f"Correlation between discounted_price and rating: {correlation_coefficient:.2f}")
print(f"Correlation between actual_price and rating:       {corr_actual:.2f}")
print(f"Correlation between discount_percentage and rating: {corr_discount_pct:.2f}")
print(f"Correlation between discount_amount and rating:    {corr_discount_amt:.2f}")
```
